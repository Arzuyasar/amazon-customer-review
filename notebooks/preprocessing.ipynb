```python
df = pd.read_csv("data/labeled_data.csv")
print(df.shape)
df.head()
```

```python
import re

def clean_text(text):
    if not isinstance(text, str):
        return ""
    text = re.sub(r"<.*?>", " ", text)      # HTML tagları kaldır
    text = re.sub(r"[^a-zA-Z\s]", " ", text) # Noktalama kaldır
    text = text.lower().strip()              # Küçük harf
    text = re.sub(r"\s+", " ", text)         # Fazla boşluk kaldır
    return text

df["clean_body"] = df["review_body"].apply(clean_text)
df["clean_headline"] = df["review_headline"].apply(clean_text)

df[["review_body", "clean_body"]].head(3)
```

```python
from sklearn.preprocessing import LabelEncoder

le = LabelEncoder()
df["label"] = le.fit_transform(df["problem_category"])

print(dict(zip(le.classes_, le.transform(le.classes_))))
df["label"].value_counts()
```
