```python
import pandas as pd
existing = pd.read_csv('/home/ece/code/arcasoyece/amazon-customer-review/data/combined_clean.csv')
print("Mevcut:", len(existing))
```

```python
# Yeni veri — her kategoriden 250k
dfs = []
for file, n in [
    ('/home/ece/code/arcasoyece/amazon-customer-review/data/amazon_reviews_us_Electronics_v1_00.tsv', 250000),
    ('/home/ece/code/arcasoyece/amazon-customer-review/data/amazon_reviews_us_Video_Games_v1_00.tsv', 250000),
    ('/home/ece/code/arcasoyece/amazon-customer-review/data/amazon_reviews_us_Mobile_Electronics_v1_00.tsv', 250000),
]:
    df = pd.read_csv(file, sep='\t', on_bad_lines='skip',
                     usecols=['review_headline','review_body','star_rating',
                              'verified_purchase','helpful_votes','total_votes'],
                     nrows=300000)
    df = df.dropna(subset=['review_body'])
    df = df.sample(n=min(n, len(df)), random_state=123)
    dfs.append(df)

new_data = pd.concat(dfs)
print("Yeni:", len(new_data))

# Birleştir, duplicate temizle
combined = pd.concat([existing, new_data]).drop_duplicates(subset=['review_body'])
combined = combined.reset_index(drop=True)
print("Toplam:", len(combined))

combined.to_csv('/home/ece/code/arcasoyece/amazon-customer-review/data/combined_clean.csv', index=False)
print("Kaydedildi.")
```
